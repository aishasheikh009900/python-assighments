```python
# Task 15 - Clinic Appointment Manager

patients = []



def add_patient():

    patient_id = int(input("Enter patient ID: "))


    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient already exists!")
            return

    patient_name = input("Enter patient name: ")
    patient_age = int(input("Enter patient age: "))
    problem = input("Enter patient's problem: ")
    time = input("Enter appointment time: ")
    emergency = input("Is emergency patient? (yes/no): ")

    patient = {
        "id": patient_id,
        "name": patient_name,
        "age": patient_age,
        "problem": problem,
        "time": time,
        "emergency": emergency
    }

    patients.append(patient)

    print("Patient added successfully!")



def search_patient():

    patient_id = int(input("Enter patient ID to search: "))

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient Found!")
            print(patient)
            return

    print("Patient not found!")


def display_patients():

    if len(patients) == 0:
        print("No patients found!")
        return

    print("\n===== Patients =====")


    for patient in patients:
        if patient["emergency"].lower() == "yes":
            print(patient)


    for patient in patients:
        if patient["emergency"].lower() == "no":
            print(patient)



def cancel_appointment():

    patient_id = int(input("Enter patient ID to cancel appointment: "))

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Appointment cancelled successfully!")
            return

    print("Patient not found!")



def count_patients():

    emergency_count = 0
    normal_count = 0

    for patient in patients:

        if patient["emergency"].lower() == "yes":
            emergency_count += 1
        else:
            normal_count += 1

    print("Normal patients:", normal_count)
    print("Emergency patients:", emergency_count)
    print("Total patients:", len(patients))



while True:

    print("\n===== Clinic Appointment Manager =====")
    print("1. Add Patient")
    print("2. Search Patient")
    print("3. Display Patients")
    print("4. Cancel Appointment")
    print("5. Count Patients")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_patient()

    elif choice == "2":
        search_patient()

    elif choice == "3":
        display_patients()

    elif choice == "4":
        cancel_appointment()

    elif choice == "5":
        count_patients()

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid choice!")
```
